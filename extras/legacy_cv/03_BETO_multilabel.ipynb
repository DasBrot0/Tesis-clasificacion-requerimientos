# BETO multietiqueta — Cross-Validation
Este notebook ejecuta el protocolo compartido de 5-fold iterative stratification. El test congelado solo se usa si `RUN_FINAL_TEST=True`.


```python
import subprocess, sys
RUN_HPARAM_SEARCH = False
USE_TRAIN_AUGMENTATION = False
USE_MODERATE_POS_WEIGHT = False
RUN_FINAL_TEST = False  # activar una sola vez, al cerrar el experimento
args = [sys.executable, 'security_cross_validation.py', '--model', 'beto']
if RUN_HPARAM_SEARCH: args.append('--hparam-search')
if USE_TRAIN_AUGMENTATION: args.append('--use-augmentation')
if USE_MODERATE_POS_WEIGHT: args.append('--moderate-pos-weight')
if RUN_FINAL_TEST: args.append('--final-test')
subprocess.run(args, check=True)
```
